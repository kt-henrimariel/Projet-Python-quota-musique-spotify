```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from itertools import combinations
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import squareform
```

```python
# =====================================================================
#  STYLE COMMUN DES GRAPHIQUES (matplotlib)
# =====================================================================
COULEURS = ["#2a78d6", "#eb6834", "#1baf7a", "#eda100",
            "#e87ba4", "#008300", "#6250d6", "#e34948"]
plt.rcParams.update({
    "axes.prop_cycle": plt.cycler(color=COULEURS),
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.alpha": 0.2, "font.size": 10,
})


def Courbes(df, colonnes, x="semaine", ncols=2, titre="", ylabel="", noms=None,
                     sharey=True, ylim=None, ligne=None, moyenne=False, bande=None,
                     calendrier=None, echelle=1, marker=None, xticks=None, note=None, rotation=45):
    """Un graphique par pays (ordre de pays_liste).

    df         : une ligne par (pays, x) ; 'pays' en colonne ou dans l'index
    colonnes   : colonne(s) tracée(s) dans chaque panneau
    echelle    : facteur multiplicatif (100 pour passer d'une part à un pourcentage)
    bande      : {colonne: colonne_borne} pour colorier l'écart entre une courbe et sa borne
    calendrier : si fourni, chaque pays est étendu à toutes ces dates (les semaines
                 manquantes restent vides, sans relier les points de part et d'autre)
    ligne      : ordonnée d'une ligne de référence ; moyenne=True trace la moyenne du pays
    """
    colonnes = [colonnes] if isinstance(colonnes, str) else list(colonnes)
    noms = noms or {}
    d = df if "pays" in df.columns else df.reset_index()
    lignes = -(-len(pays_liste) // ncols)
    fig, axes = plt.subplots(lignes, ncols, figsize=(6 * ncols, 2.3 * lignes + 1),
                             sharex=True, sharey=sharey, squeeze=False)
    for axe, pays in zip(axes.flat, pays_liste):
        dp = d[d["pays"] == pays].set_index(x).sort_index()
        if calendrier is not None:
            dp = dp.reindex(calendrier)
        for c in colonnes:
            axe.plot(dp.index, dp[c] * echelle, marker=marker, label=noms.get(c, c))
            if bande:
                axe.fill_between(dp.index, dp[c] * echelle, dp[bande[c]] * echelle, alpha=0.2)
        if moyenne:
            axe.axhline((dp[colonnes[0]] * echelle).mean(), color="gray", linestyle="--")
        if ligne is not None:
            axe.axhline(ligne, color="gray", linestyle="--")
        axe.set_title(pays, fontsize=10)
        if ylim:
            axe.set_ylim(ylim)
        if xticks is not None:
            axe.set_xticks(xticks)
        axe.tick_params(axis="x", labelrotation=rotation)
        for etiquette in axe.get_xticklabels():
            etiquette.set_ha("right" if rotation else "center")
    for axe in axes.flat[len(pays_liste):]:
        axe.set_visible(False)
    if len(colonnes) > 1:
        axes.flat[0].legend(fontsize=8)
    fig.supylabel(ylabel)
    fig.suptitle(titre)
    if note:
        fig.text(0.5, 0.005, note, ha="center", fontsize=8)
    fig.tight_layout(rect=(0, 0.02 if note else 0, 1, 0.98))
    plt.show()
    plt.close(fig)


def carte_chaleur(m, titre, figsize=(8, 6.5), vmin=None, vmax=None, fmt=".1f",
                  label="%", labels_y=None):
    """Carte de chaleur d'un tableau (DataFrame) avec la valeur écrite dans chaque case.
    Les cases vides (NaN) restent blanches."""
    fig, ax = plt.subplots(figsize=figsize)
    image = ax.imshow(m, cmap="Blues", aspect="auto", vmin=vmin, vmax=vmax)
    ax.set_xticks(range(len(m.columns)), m.columns, rotation=45, ha="right")
    ax.set_yticks(range(len(m.index)), labels_y if labels_y is not None else m.index)
    ax.grid(False)
    seuil = 0.6 * np.nanmax(m.to_numpy())
    for i in range(m.shape[0]):
        for j in range(m.shape[1]):
            if not np.isnan(m.iloc[i, j]):
                ax.text(j, i, format(m.iloc[i, j], fmt), ha="center", va="center", fontsize=8,
                        color="white" if m.iloc[i, j] > seuil else "black")
    fig.colorbar(image, ax=ax, label=label)
    ax.set_title(titre, fontsize=10)
    fig.tight_layout()
    plt.show()
    plt.close(fig)


def valeurs_dans_barres(ax, fmt="{:.1f}", horizontal=True):
    """Écrit la valeur de chaque barre en son centre (ou au centre de chaque modalité
    d'une barre empilée). Le texte est blanc sur fond foncé, noir sur fond clair."""
    echelle_log = ax.get_xscale() == "log" if horizontal else ax.get_yscale() == "log"
    for barre in ax.patches:
        if horizontal:
            debut, taille = barre.get_x(), barre.get_width()
            fin = debut + taille
            if echelle_log:
                debut = max(debut, ax.get_xlim()[0])
                milieu = (debut * fin) ** 0.5
            else:
                milieu = debut + taille / 2
            x, y, valeur = milieu, barre.get_y() + barre.get_height() / 2, taille
        else:
            x, y, valeur = barre.get_x() + barre.get_width() / 2, barre.get_y() + barre.get_height() / 2, barre.get_height()
        r, g, b, _ = barre.get_facecolor()
        clair = 0.299 * r + 0.587 * g + 0.114 * b > 0.6
        ax.text(x, y, fmt.format(valeur), ha="center", va="center", fontsize=9,
                color="black" if clair else "white")
```

Pendant longtemps, la musique entendue dans un pays dépendait de ce que les radios, les maisons de disques et les disquaires choisissaient de rendre disponible. Ces intermédiaires étaient en grande partie nationaux : ils programmaient surtout des artistes locaux et les succès étrangers jugés adaptés à leur public. Plusieurs pays ont encadré cette programmation par des **quotas de diffusion**, c'est-à-dire l'obligation de réserver une part du temps d'antenne à certains contenus. La circulation des musiques d'un pays à l'autre existait, mais elle passait par ces filtres.

La mondialisation des échanges culturels, puis l'arrivée d'internet, ont profondément modifié ce cadre. La diffusion d'un titre ne dépend plus de l'accord d'un programmateur national : un même catalogue de plusieurs dizaines de millions de chansons est accessible partout, à toute heure, depuis un téléphone. Les plateformes de streaming comme Spotify occupent une place croissante à côté de la radio, et leurs playlists et leurs recommandations jouent un rôle comparable à celui d'un programmateur, mais à l'échelle mondiale. On pourrait en conclure que les auditeurs du monde entier écoutent progressivement les mêmes artistes. Pourtant, partager une plateforme ne signifie pas partager les mêmes goûts : la langue des chansons, l'origine des artistes et les habitudes d'écoute peuvent maintenir de fortes différences entre pays.

C'est dans ce contexte que la question des quotas se pose. Conçus pour la radio, ils subsistent alors que l'écoute se déplace vers le streaming, et leur effet sur la musique réellement écoutée reste discuté. Ils sont en outre de natures différentes. Certains portent sur l'**origine des artistes** (l'Argentine impose 30 % de musique d'origine nationale), d'autres sur la **langue des chansons** (la France impose 40 % de chansons d'expression française, le Portugal 30 % de musique portugaise définie par la langue ou le patrimoine culturel). Ce travail pose donc la question suivante : **les pays qui imposent des quotas de diffusion ont-ils, dans les classements Spotify, une répartition des écoutes plus favorable aux artistes nationaux et aux chansons en langue locale que les pays qui n'en imposent pas ?** 

La réponse à cette question s'articulera autour du plan suivant. La revue de littérature précise le cadre et les questions. La partie <a href="#s3">3</a> décrit le volume, la dynamique et le renouvellement des classements. La partie <a href="#s4">4</a> étudie leur structure, avec la concentration des écoutes puis la ressemblance entre pays. La partie <a href="#s5">5</a> mesure la place de la musique nationale et de la langue locale et la compare selon la présence et la nature des quotas. La partie <a href="#s6">6</a> discute ces résultats et leurs limites, puis la conclusion répond à la question posée. L'analyse reste descriptive : elle décrit des écarts entre pays sans prétendre mesurer l'effet causal d'une politique.

L’accès à un catalogue musical international pourrait laisser penser que les auditeurs écoutent progressivement les mêmes artistes, indépendamment de leur pays. Pourtant, partager une plateforme ne signifie pas nécessairement partager les mêmes goûts. Les préférences peuvent rester liées à la langue et à l’origine des artistes, tandis que les radios et les plateformes influencent la visibilité des chansons. Notre projet s’inscrit dans ce débat : quelle place les répertoires nationaux occupent-ils dans les classements Spotify, et comment interpréter les différences entre pays, notamment au regard des quotas de diffusion ?

### Des échanges internationaux qui n’effacent pas les préférences nationales

Ferreira et Waldfogel (2013) étudient la circulation internationale de la musique à partir des classements de vingt-deux pays sur plusieurs décennies. Ils mettent en évidence un « biais domestique » : la musique d’un pays rencontre un succès particulièrement important sur son propre marché. Ce résultat remet en cause l’idée selon laquelle l’ouverture aux productions étrangères conduirait nécessairement à l’effacement des répertoires locaux. Les auteurs constatent même un renforcement de ce biais durant la dernière décennie de leur période d’étude. La proximité géographique et linguistique contribue également à expliquer la circulation des musiques entre pays.

Bello et Garcia (2021) prolongent cette réflexion à l’ère du streaming. En étudiant les classements de trente-neuf pays sur quatre ans, à partir de Spotify et d’iTunes, ils observent une différenciation croissante entre les classements nationaux. Leur résultat va donc à l’encontre d’une convergence systématique vers un petit ensemble de succès mondiaux. Il faut toutefois distinguer cette diversité **entre pays** de la diversité **à l’intérieur d’un pays** : des classements nationaux peuvent être très différents tout en étant chacun dominés par quelques artistes.

Ces travaux justifient deux dimensions complémentaires de notre analyse. La comparaison des titres présents dans plusieurs pays permet d’étudier la ressemblance entre classements. La répartition des écoutes entre chansons et artistes permet, elle, de mesurer leur concentration. Une forte présence de musique nationale ne suffit donc pas à conclure que les écoutes sont diversifiées.

### Les quotas favorisent-ils les répertoires nationaux et leur diversité ?

Les quotas de diffusion cherchent à réserver une place à certains contenus dans la programmation. Leur justification repose notamment sur l’idée qu’une exposition plus importante peut soutenir les artistes concernés et favoriser leur découverte. Mais augmenter la présence d’un répertoire et diversifier les œuvres diffusées sont deux objectifs distincts.

Perona (2011) montre, dans un modèle théorique, qu’un quota peut inciter un diffuseur à répéter davantage les contenus nationaux les plus populaires afin de conserver son audience. Cette répétition peut alors réduire la place disponible pour des productions moins connues. Ce mécanisme ne démontre pas que tous les quotas réduisent effectivement la diversité ; il souligne que leur effet dépend aussi de la réaction des diffuseurs. Pour notre projet, cela motive l’étude du poids des artistes les plus écoutés : une part nationale élevée peut correspondre au succès de nombreux artistes ou à la domination de quelques vedettes.

Le document de travail de Bacache, Louail, Renault et Senik (2026) examine plus directement le lien entre quotas et streaming. Les auteurs rapprochent les politiques de quotas des classements Spotify de soixante et onze pays et mobilisent également des données individuelles de Deezer en France. Leurs analyses ne mettent pas en évidence de retombées substantielles des quotas sur la part de musique nationale consommée en streaming. Ils observent notamment une part d’artistes nationaux plus faible dans les pays avec quotas, mais insistent sur la difficulté d’interprétation : ces politiques peuvent précisément avoir été adoptées dans des marchés où le répertoire national était peu représenté.

Cette étude invite à distinguer association et causalité. Un écart entre pays avec et sans quotas ne mesure pas, à lui seul, l’effet de la réglementation. Les pays diffèrent également par leur langue, leur industrie musicale et leurs habitudes d’écoute. De même, ne pas détecter d’effet dans les analyses réalisées ne prouve pas que les quotas sont sans conséquence dans tous les contextes. Ces précautions sont particulièrement importantes pour notre comparaison, qui porte sur un nombre limité de pays.

### Le rôle des plateformes dans les écoutes observées

Les écoutes en streaming ne dépendent pas uniquement des préférences préexistantes des auditeurs ou de leur exposition à la radio. Aguiar et Waldfogel (2021) étudient la promotion musicale sur Spotify et mettent en évidence des effets importants de l’inclusion dans les playlists sur le succès des chansons. La visibilité accordée par une plateforme constitue donc un mécanisme à prendre en compte lorsqu’on interprète les classements : les titres les plus écoutés ne reflètent pas seulement des choix effectués indépendamment de toute mise en avant.

Matrosova et ses coauteurs (2024) analysent plus spécifiquement la place de la musique locale dans les recommandations. En confrontant plusieurs jeux de données et différents modèles, ils montrent que les conclusions dépendent des utilisateurs étudiés, des paramètres des algorithmes et de la manière d’identifier la musique locale. On ne peut donc pas considérer que les recommandations favorisent systématiquement les productions nationales, ou systématiquement les productions étrangères. Leur étude souligne aussi l’importance de la qualité des informations sur l’origine des artistes.

Cette question rejoint directement la construction de notre base. La nationalité d’un artiste et la langue d’une chanson doivent être distinguées : un artiste national peut chanter dans une langue étrangère, et un artiste étranger dans la langue du pays étudié. Les collaborations internationales compliquent également la classification. Les choix d’enrichissement et de traitement de ces situations influencent donc les parts mesurées et doivent être explicités.

### La place de notre étude

Ces recherches conduisent à examiner ensemble la présence des répertoires nationaux, la concentration des écoutes et la ressemblance entre pays. Notre analyse des classements hebdomadaires Spotify permet de suivre ces dimensions dans le temps, puis d’étudier les liens entre succès des titres, origine des artistes et langue chantée. La comparaison selon les quotas, qui constitue la question centrale de ce travail, est menée sans prétendre à une évaluation causale de leur efficacité.

Enfin, notre observation porte sur les chansons du Top 200, et non sur l’ensemble des écoutes ou sur les goûts de chaque utilisateur. Les résultats décrivent donc les succès les plus visibles dans les pays étudiés. La présente étude repose sur les classements hebdomadaires Spotify Top 200 de dix pays : *l'Argentine, l'Allemagne, l'Égypte, la France, la Grèce, le Japon, la Corée du Sud, le Maroc, le Portugal et la Thaïlande*. La période d'étude couvre **210** semaines, de la semaine du *06/10/2022* à celle du *08/10/2026*. Chaque observation correspond à un titre présent dans le classement d'un pays au cours d'une semaine donnée. Les informations disponibles comprennent notamment le rang occupé par le titre et son nombre d'écoutes (streams).

#### Collecte

Les données ont été recueillies directement sur la plateforme Spotify Charts. Les conditions d'utilisation de Spotify ne permettant pas de procéder à une extraction automatisée des classements par scraping, la collecte a été réalisée par téléchargement individuel des fichiers hebdomadaires. Pour chacun des dix pays, les classements ont ainsi été téléchargés sur les 210 semaines de la période étudiée, conduisant à la constitution de 2 100 fichiers.

Ces fichiers ont ensuite été regroupés dans une base de données unique, dénommée base_complete, afin de disposer d'un ensemble de données exploitable pour les analyses statistiques. Chaque observation a été associée à son pays et à sa semaine de référence.

#### Vérification

Une première vérification a porté sur la cohérence des fichiers téléchargés. Les noms des variables étant identiques dans l'ensemble des fichiers, aucune harmonisation supplémentaire n'a été nécessaire. De même, aucun doublon n'a été relevé après leur regroupement.

Compte tenu des dix pays et des 210 semaines retenus, le volume théorique de la base était 420 000 observations. Toutefois, la base constituée comporte 419 891 lignes. L'examen de cet écart a permis de constater que certaines observations étaient absentes des fichiers disponibles, en raison notamment du retrait de certains artistes des classements Spotify.

Les classements Spotify ne fournissant pas d'informations sur la **nationalité des artistes** ni sur la **langue des chansons**, un travail complémentaire d'enrichissement a été nécessaire. Ces deux dimensions étant au cœur de la problématique étudiée, leur identification constituait une étape essentielle de la préparation des données.

La première source mobilisée a été **<u>MusicBrainz</u>**, une base de données musicale ouverte qui rassemble des informations sur les artistes, les groupes et les œuvres musicales. Un programme développé en Python a permis d'effectuer des recherches automatisées dans cette base afin de récupérer les informations disponibles sur les artistes et les titres présents dans notre échantillon.

L'enrichissement a été réalisé à partir des valeurs uniques plutôt que sur l'ensemble des observations, afin d'éviter de répéter les mêmes recherches pour un artiste ou un titre apparaissant plusieurs fois dans les classements. Ainsi, la recherche des langues a porté sur **20 392 couples distincts (artiste, titre)**, tandis que celle des nationalités a concerné **9 843 artistes distincts**.

Les nationalités sont codées selon la norme ISO 3166-1 alpha-2, qui associe à chaque pays un code de deux lettres. Lorsqu'un artiste ou un groupe est associé à plusieurs nationalités, les différents codes sont conservés et séparés par un point-virgule (par exemple FR;MA).

Les langues sont quant à elles codées selon la norme ISO 639-3, qui attribue un code de trois lettres à chaque langue. On retrouve ainsi les codes eng pour l'anglais, fra pour le français ou encore kor pour le coréen. Lorsqu'une chanson comporte plusieurs langues, les différents codes sont conservés et séparés par une virgule.

Toutefois, MusicBrainz ne renseigne pas systématiquement toutes les informations recherchées : à l'issue de cette première étape, la langue restait manquante pour 1 453 titres (7,1 %) et la nationalité pour 483 artistes (4,9 %). Pour compléter ces informations, nous avons eu recours à *Perplexity*, un outil d'intelligence artificielle capable d'effectuer des recherches sur Internet à partir de différentes sources. Les informations obtenues ont ensuite été examinées à l'aide de *Claude*, utilisé comme outil complémentaire de vérification, notamment pour les informations relatives aux artistes et à leur environnement musical.

Cette démarche a permis de compléter la quasi-totalité des informations initialement manquantes : il ne reste que 2 langues et 6 nationalités non renseignées. Les résultats ont ensuite été intégrés dans base_complete, à travers deux nouvelles colonnes : la nationalité des artistes et la langue des chansons.

Compte tenu de l'importance des variables de nationalité et de langue dans cette étude, plusieurs vérifications ont été effectuées afin d'évaluer la cohérence des informations ajoutées et de limiter les erreurs de classification.

**Vérification des valeurs** 

Les informations recueillies auprès de Perplexity ont été examinées par Claude, utilisé comme outil de vérification. Par ailleurs, un échantillon aléatoire de 50 titres parmi les 1 453 renseignés à l'aide de l'intelligence artificielle a été constitué et vérifié à partir de recherches sur Internet. La même démarche a été appliquée à 20 artistes. Dans les deux cas, les informations fournies se sont révélées exactes, soit un taux de conformité de 100 %.

Une attention particulière a également été accordée aux valeurs présentant des caractéristiques inhabituelles : langues rarement rencontrées dans l'échantillon, nationalités semblant incompatibles avec les autres informations disponibles ou codes linguistiques atypiques. Ces cas ont été vérifiés individuellement afin de distinguer les situations réelles des erreurs d'attribution.

**Traçabilité** 

Afin de garantir la transparence du processus d'enrichissement, les informations ajoutées en dehors de MusicBrainz ont été conservées dans deux fichiers distincts, `ajouts_langues.csv` (1 451 titres) et `ajouts_nationalites.csv` (477 artistes). Chaque ligne indique l'origine de la valeur ainsi que sa source ou son fondement. 

**Complétude finale** 

À l'issue du processus, 2 langues (0,01 % des 20 392 titres) et 6 nationalités (0,06 % des 9 843 artistes) restent non renseignées. Elles sont laissées vides plutôt que remplacées par une modalité « indéterminé », et sont donc traitées comme des valeurs non nationales et non locales dans la construction des variables. Elles représentent une part négligeable de la base.

<h2 style="background:#eaf2fb;color:#12304d;padding:10px 18px;border-left:7px solid #eb6834;border-radius:0 8px 8px 0;font-size:21px;font-family:Georgia,serif;margin:22px 0 8px 0;"><a id="s2-4"></a>2.4. CONSTRUCTION DES VARIABLES</h2>


Les colonnes `nationalite` et `langue` servent à construire, pour chaque observation, trois indicateurs binaires utilisés dans les analyses :

- **National** : vaut 1 si au moins une origine renseignée correspond au pays du classement. Une collaboration internationale compte donc comme nationale dès qu'une de ses origines est locale (un titre d'Aya Nakamura est national en France, mais pas en Allemagne).
- **LangueLocale** : vaut 1 si au moins une langue de la chanson appartient à la liste des langues locales du pays d'écoute (langue principale et variantes, par exemple l'arabe égyptien pour l'Égypte).
- **Anglais** : vaut 1 si l'anglais figure parmi les langues de la chanson.

L'origine porte sur les artistes et la langue sur la chanson : les deux indicateurs peuvent donc différer pour un même pays. Un titre bilingue peut être à la fois en langue locale et en anglais, de sorte que ces parts ne s'additionnent pas. Les origines ou langues non renseignées donnent la valeur 0 et sont repérées par deux indicatrices (`origine_connue` et `langue_connue`). Enfin, les crédits d'artistes sont séparés en une liste d'artistes, dont le premier (`artiste`) est retenu comme artiste principal.

```python
# Valeurs manquantes : chaîne vide
base["nationalite"] = base["nationalite"].fillna("")
base["langue"] = base["langue"].fillna("")

# Code ISO des artistes (en minuscules) et langues considérées comme locales dans chaque pays
codes_pays = {
    'Allemagne': 'de', 'Argentine': 'ar', 'Corée du Sud': 'kr', 'Égypte': 'eg',
    'France': 'fr', 'Grèce': 'gr', 'Japon': 'jp', 'Maroc': 'ma',
    'Portugal': 'pt', 'Thaïlande': 'th'
}
langues_locales = {
    'Allemagne': ['deu', 'ger'], 'Argentine': ['spa'], 'Corée du Sud': ['kor'],
    'Égypte': ['ara', 'arb', 'arz'], 'France': ['fra', 'fre'], 'Grèce': ['ell', 'gre'],
    'Japon': ['jpn'], 'Maroc': ['ara', 'arb', 'ary', 'ber', 'zgh', 'tzm', 'shi', 'rif'],
    'Portugal': ['por'], 'Thaïlande': ['tha']
}

def lire_codes(valeur):
    """'FR;BE' ou 'fra, eng' -> liste de codes en minuscules, sans doublon ni valeur inconnue."""
    texte = str(valeur).lower().replace(';', ',').replace(' ', ',')
    inconnus = ['', 'nan', 'mul', 'und', 'unknown', 'indetermine',
                'indéterminé', 'inconnue', 'inconnu', 'none']
    resultat = []
    for code in texte.split(','):
        if code not in inconnus and code not in resultat:
            resultat.append(code)
    return resultat

def separer_artistes(credit):
    """'A, B' -> ['A', 'B'], en protégeant les noms qui contiennent eux-mêmes une virgule."""
    credit = credit.replace('Tyler, The Creator', 'Tyler__The Creator')
    credit = credit.replace('Earth, Wind & Fire', 'Earth__Wind & Fire')
    if credit == 'Niaks & Maes':
        credit = 'Niaks, Maes'
    if credit == 'TIF & Flenn':
        credit = 'TIF, Flenn'
    noms = []
    for nom in credit.split(','):
        nom = nom.strip().replace('__', ', ')
        if nom and nom not in noms:
            noms.append(nom)
    return noms

# Listes de codes et d'artistes
base['origines'] = base['nationalite'].apply(lire_codes)
base['langues'] = base['langue'].apply(lire_codes)
base['artistes'] = base['artist_names'].apply(separer_artistes)
base['artiste'] = base['artistes'].str[0].astype('string')       # artiste principal : premier crédit

# Indicateurs binaires
base['National'] = [int(codes_pays[p] in o) for p, o in zip(base['pays'], base['origines'])]
base['LangueLocale'] = [int(any(l in langues_locales[p] for l in ls))
                        for p, ls in zip(base['pays'], base['langues'])]
base['Anglais'] = base['langues'].apply(lambda ls: int('eng' in ls))
base['origine_connue'] = base['origines'].str.len() > 0
base['langue_connue'] = base['langues'].str.len() > 0

print(base[['National', 'LangueLocale', 'Anglais']].mean() * 100)
```

Ainsi, 61,88 % des observations concernent des chansons d'artistes nationaux, 71,39 % des chansons chantées dans la langue locale et 24,18 % des chansons comportant de l'anglais (observations pondérées de la même façon, quel que soit le nombre d'écoutes).

<h3 style="color:#2a78d6;border-bottom:2px solid #2a78d6;padding-bottom:4px;font-size:17px;font-family:Georgia,serif;margin:18px 0 6px 0;"><a id="s3-1-2"></a>3.1.2 Saisons</h3>

Dans cette partie, il s'agit de regarder si certains mois ont plus d'écoutes que d'autres. Par exemple, est-ce que décembre est toujours un mois fort ?

On ne peut pas comparer les mois directement, parce que le total change d'une année à l'autre. Alors on compare chaque semaine à la moyenne de son année.

Les étapes :

1. On prend le total des écoutes de chaque semaine, pour chaque pays (vu en <a href="#s3-1-1">3.1.1</a>).
2. On garde seulement 2023, 2024 et 2025, car ce sont les années complètes (52 semaines). 2022 n'a que 13 semaines (octobre à décembre) et 2026 que 41 semaines (jusqu'en octobre) : leur moyenne serait faussée.
3. Pour chaque pays et chaque année, on calcule la moyenne des écoutes par semaine.
4. Pour chaque semaine, on calcule le ratio : les écoutes de la semaine divisées par la moyenne de son année. Un ratio de 1 veut dire que la semaine est dans la moyenne. Un ratio de 1,2 veut dire 20 % au-dessus. Un ratio de 0,8 veut dire 20 % en dessous.
5. Pour chaque pays et chaque mois, on fait la moyenne des ratios des semaines de ce mois, sur les trois années. Pour décembre, c'est la moyenne de 12 semaines. On a 12 nombres par pays, donc 120 en tout.
6. On trace ces 12 nombres pour chaque pays, avec un trait en pointillé à 1. On les met aussi dans un tableau.

Si le ratio d'un mois est au-dessus de 1, ce mois est plus fort que la moyenne de l'année. Par exemple si un Pays a 1,50 en décembre : ses semaines de décembre sont en moyenne 50 % au-dessus de la moyenne de l'année.

```python
totaux["annee"] = totaux["semaine"].dt.year
totaux["mois"] = totaux["semaine"].dt.month
print(totaux.groupby("annee")["semaine"].nunique())
```

```python
complets = totaux[totaux["annee"].isin([2023, 2024, 2025])].copy()

moyennes = complets.groupby(["pays", "annee"])["streams"].mean().reset_index()
moyennes = moyennes.rename(columns={"streams": "moyenne_annee"})
complets = complets.merge(moyennes, on=["pays", "annee"], how="left")

complets["ratio"] = complets["streams"] / complets["moyenne_annee"]
print(complets.shape, complets["ratio"].mean())
```

```python
saison = complets.groupby(["pays", "mois"])["ratio"].mean().reset_index()
print(saison.shape)
print(saison.head(12))
```

On trace le ratio moyen de chaque mois, pour chaque pays. La ligne en pointillé est la moyenne de l'année (ratio = 1) : au-dessus, le mois est plus fort que la moyenne ; en dessous, il est plus faible.

```python
Courbes(saison, "ratio", x="mois", ncols=2, ylim=(0.6, 1.4), ligne=1, marker="o", xticks=range(1, 13), rotation=0,
                 titre="Ratio moyen de chaque mois à la moyenne de l'année", ylabel="Ratio (1 = moyenne de l'année)")
```

Les mêmes valeurs en tableau, pour lire les chiffres exacts (un pays par ligne, un mois par colonne).

```python
tableau = saison.pivot(index="pays", columns="mois", values="ratio").round(2)
print(tableau)
```

Sur les courbes de <a href="#s3-1-1">3.1.1</a>, on voit parfois un pic : le total d'un pays saute d'un coup, puis redescend. On veut savoir quelles semaines sont dans ce cas et ce qui les explique.

C'est utile parce que ces semaines changent fortement le total : elles peuvent fausser les moyennes et les comparaisons entre pays.

Pour repérer une semaine exceptionnelle, on n'a pas fixé de seuil. On calcule la variation de chaque semaine par rapport à la semaine d'avant, en pourcentage, et on garde les 10 plus grosses hausses. On regarde seulement les hausses.

Les étapes :

1. On calcule la variation de chaque semaine, pays par pays.
2. On affiche les 10 plus grosses hausses.
3. On regarde quels pays reviennent le plus souvent, et on étudie ceux-là.
4. Pour chaque grosse semaine, on affiche les 10 chansons les plus écoutées, pour voir ce qui domine.
5. Si un seul artiste semble dominer, on calcule sa part dans le total de la semaine.

```python
totaux["variation"] = totaux.groupby("pays")["streams"].pct_change()
print(totaux.sort_values("variation", ascending=False).head(10))
```

Dans ces 10 semaines, l'Allemagne revient 3 fois, le Maroc 4 fois, la Grèce 2 fois et l'Égypte 1 fois. Les sauts vont de +36 % à +78 %. On étudie l'Allemagne, le Maroc et la Grèce, qui font 9 des 10 semaines. Pour l'Allemagne, on prend ses trois semaines de fin décembre. Pour le Maroc et la Grèce, on prend leur plus gros saut.

```python
# Les 10 chansons les plus écoutées d'un pays une semaine donnée
def top_titres(pays, semaine, n=10):
    sp = base[(base["pays"] == pays) & (base["semaine"] == semaine)]
    return sp.sort_values("streams", ascending=False)[["artist_names", "track_name", "streams"]].head(n)

top_titres("Allemagne", "2025-12-25")
```

STRUCTURE DES CLASSEMENTS : CONCENTRATION ET RESSEMBLANCE <a href="#plan" style="float:right;font-size:12px;color:#dbe8f7;text-decoration:none;font-family:sans-serif;">↑ retour au plan</a></h1>


La partie <a href="#s3">3</a> a décrit l'évolution des classements dans le temps : volume des écoutes, saisons, renouvellement des titres. Cette partie en décrit la **structure**, d'abord à l'intérieur de chaque classement, puis entre classements. La concentration (<a href="#s4-1">4.1</a>) mesure si les écoutes d'un pays se répartissent sur beaucoup de titres et d'artistes ou sur quelques-uns. La ressemblance (<a href="#s4-2">4.2</a>) mesure la part des titres et des écoutes que deux pays ont en commun, puis regroupe les pays selon cette ressemblance. Ces mesures portent sur les classements entiers et ne font pas intervenir la nationalité ni la langue des titres.

<h2 style="background:#eaf2fb;color:#12304d;padding:10px 18px;border-left:7px solid #eb6834;border-radius:0 8px 8px 0;font-size:21px;font-family:Georgia,serif;margin:22px 0 8px 0;"><a id="s4-1"></a>4.1. CONCENTRATION DU MARCHE</h2>


Cette partie examine si les écoutes se concentrent sur quelques titres ou quelques artistes. Pour chaque pays et chaque semaine, on étudie la répartition des écoutes du Top 200, d'abord entre les chansons, puis entre les artistes : plusieurs chansons pouvant appartenir au même artiste, un classement varié en titres peut être dominé par quelques artistes.

Un classement incomplet fausserait les proportions, puisque les écoutes d'une chanson manquante manquent aussi au total. On ne retient donc que les semaines comptant 200 titres et 200 rangs distincts, après avoir vérifié l'absence de doublons et de valeurs manquantes. Dans le tableau, `sum` est le nombre de semaines complètes et `count` le nombre de semaines disponibles.

```python
cles = ['pays', 'semaine']

colonnes = cles + ['uri', 'rank', 'streams', 'artist_names']
assert not base[colonnes].isna().any().any(), 'Valeurs manquantes à examiner'
assert not base.duplicated(cles + ['uri']).any(), 'Titres en double'
assert not base.duplicated(cles + ['rank']).any(), 'Rangs en double'
assert base['streams'].gt(0).all()
assert base['rank'].between(1, 200).all()

controle = base.groupby(cles).agg(lignes=('uri', 'size'), rangs=('rank', 'nunique'))
controle['complet'] = (controle['lignes'] == 200) & (controle['rangs'] == 200)
base = base.merge(controle['complet'], on=cles, validate='many_to_one')
donnees = base[base['complet']].copy()
print(controle.groupby('pays')['complet'].agg(['sum', 'count']).to_string())
print('Classements exclus :', (~controle['complet']).sum())

# Semaines où les 10 pays ont un classement complet (comparaisons à périmètre identique)
nombre_pays = controle['complet'].groupby('semaine').sum()
semaines_communes = nombre_pays[nombre_pays == len(pays_liste)].index
print('Semaines complètes communes :', len(semaines_communes))
```

On conserve 2 000 classements sur 2 100 (95,2 %). Cela améliore la comparabilité sans prouver que les semaines exclues sont aléatoires ; l'effet de cette exclusion est vérifié en <a href="#s4-1-3">4.1.3</a>.

<h3 style="color:#2a78d6;border-bottom:2px solid #2a78d6;padding-bottom:4px;font-size:17px;font-family:Georgia,serif;margin:18px 0 6px 0;"><a id="s4-1-1"></a>4.1.1. Mesurer la concentration par chanson et par artiste</h3>


Quatre indicateurs sont calculés pour chaque pays et chaque semaine.

* **Par chanson.** 
    - La **part du Top 10** est le rapport entre les écoutes des dix premières chansons et celles des 200 chansons (avec 200 titres également écoutés, elle vaudrait 5 %). 
    - Le **nombre de titres pour atteindre 50 %** des écoutes s'obtient en descendant le classement : plus il est faible, plus la tête du classement concentre les écoutes. 
    - Le **HHI** est la somme des parts de chaque chanson au carré, c'est-à-dire la probabilité que deux écoutes tirées au hasard portent sur la même chanson (0,005 si les 200 titres sont également écoutés). Son inverse, le **nombre effectif de titres** (1 / HHI), s'interprète ainsi : un résultat de 80 signifie une concentration équivalente à 80 titres de même poids, même si 200 sont présents.

* **Par artiste.** 
    - La **part du Top 5 des artistes** s'obtient en additionnant d'abord toutes les chansons de chaque artiste. Pour une collaboration, les écoutes sont réparties également : 900 écoutes avec trois artistes donnent 300 à chacun, et le total reste 900.


```python
donnees = donnees.sort_values(cles + ['rank'])
donnees['part'] = donnees['streams'] / donnees.groupby(cles)['streams'].transform('sum')
donnees['part_cumulee'] = donnees.groupby(cles)['part'].cumsum()

titres = donnees.groupby(cles).agg(nombre_titres=('uri', 'size'))
titres['top1'] = donnees[donnees['rank'] == 1].groupby(cles)['part'].sum()
titres['top5'] = donnees[donnees['rank'] <= 5].groupby(cles)['part'].sum()
titres['top10'] = donnees[donnees['rank'] <= 10].groupby(cles)['part'].sum()
titres['titres_pour_50'] = donnees[donnees['part_cumulee'] < 0.5].groupby(cles).size().reindex(titres.index, fill_value=0) + 1

# Le HHI additionne les parts au carré ; son inverse donne un nombre effectif de titres.
donnees['part_carree'] = donnees['part'] ** 2
titres['hhi'] = donnees.groupby(cles)['part_carree'].sum()
titres['titres_effectifs'] = 1 / titres['hhi']
print(titres.head().round(3).to_string())
```

```python
# Chaque chanson est répartie à parts égales entre ses artistes (donnees['artistes'], construit en 2.4)
credits = donnees[cles + ['streams', 'artistes']].copy()
credits['ecoutes_partagees'] = credits['streams'] / credits['artistes'].str.len()
credits = credits.explode('artistes')
parts_artistes = credits.groupby(cles + ['artistes'])['ecoutes_partagees'].sum().reset_index()
parts_artistes['part'] = parts_artistes['ecoutes_partagees'] / parts_artistes.groupby(cles)['ecoutes_partagees'].transform('sum')
parts_artistes = parts_artistes.sort_values(cles + ['ecoutes_partagees'], ascending=[True, True, False])
parts_artistes['position'] = parts_artistes.groupby(cles).cumcount() + 1
top5_artistes = parts_artistes[parts_artistes['position'] <= 5].groupby(cles)['part'].sum()

# Le partage doit conserver le total des écoutes de chaque classement.
total_avant = donnees.groupby(cles)['streams'].sum()
total_apres = parts_artistes.groupby(cles)['ecoutes_partagees'].sum()
assert ((total_apres - total_avant).abs() / total_avant < 0.000000001).all()
print(parts_artistes.head().to_string(index=False))
```

Exemple pour l'Allemagne, semaine du 6 octobre 2022 : la première chanson représente 4 % des écoutes, les dix premières 18,8 %, et 52 chansons suffisent pour atteindre la moitié des écoutes (contre 100 si les 200 titres étaient également écoutés). Le nombre effectif de titres est de 119,4. Par artiste, Luciano réunit 9,48 % des écoutes avec l'ensemble de ses titres, contre 4 % pour sa première chanson, et les cinq premiers artistes réunissent 22,2 % : l'analyse par artiste fait apparaître une concentration que le poids d'une chanson isolée ne montre pas. Ces résultats dépendent de la convention de partage des collaborations.

<h3 style="color:#2a78d6;border-bottom:2px solid #2a78d6;padding-bottom:4px;font-size:17px;font-family:Georgia,serif;margin:18px 0 6px 0;"><a id="s4-1-2"></a>4.1.2. Comparer les pays et suivre l'évolution</h3>


Les parts sont calculées par semaine puis moyennées par pays, chaque semaine ayant le même poids. Le nombre de titres pour 50 % est résumé par sa médiane, c'est-à-dire la valeur centrale des semaines.

```python
resume = titres.groupby('pays').agg(
    semaines=('top10', 'size'),
    top10_titres=('top10', 'mean'),
    titres_pour_50=('titres_pour_50', 'median'),
    titres_effectifs=('titres_effectifs', 'mean'))
resume['top5_artistes'] = top5_artistes.groupby('pays').mean()
resume = resume.sort_values('top5_artistes')
affichage = resume.copy()
affichage[['top10_titres', 'top5_artistes']] *= 100
print(affichage.round(1).to_string())

ax = (resume[['top10_titres', 'top5_artistes']] * 100).plot.barh(figsize=(10, 6))
valeurs_dans_barres(ax)
ax.set_xlabel('Part moyenne des écoutes du Top 200 (%)')
ax.set_ylabel('')
ax.set_title('Concentration par titre et par artiste')
ax.legend(['10 premiers titres', '5 premiers artistes'])
plt.tight_layout()
plt.show()
plt.close()
```

Dans tous les pays, les dix premières chansons représentent entre 14,7 % et 21,2 % des écoutes du Top 200, contre 5 % si les 200 titres étaient également écoutés, et il faut en médiane entre 47 et 62 titres pour atteindre la moitié des écoutes.

Le contraste Japon-Allemagne montre l'intérêt de distinguer chansons et artistes : la part des dix premières chansons est presque identique (14,8 % et 14,9 %), mais les cinq premiers artistes réunissent 36,3 % des écoutes au Japon contre 16,2 % en Allemagne. Un classement peut donc comporter de nombreuses chansons différentes tout en concentrant ses écoutes sur quelques artistes.

Ces moyennes masquent des variations : les courbes suivantes donnent, semaine par semaine, la part du Top 10 des titres et du Top 5 des artistes. La ligne pointillée est la moyenne du pays, et les interruptions correspondent aux semaines exclues, jamais à une absence d'écoutes.

```python
hebdo = titres[['top10']].copy()
hebdo['top5_artistes'] = top5_artistes
hebdo = hebdo.reset_index()

for variable, titre in [('top10', 'Part des 10 premiers titres'),
                        ('top5_artistes', 'Part des 5 premiers artistes')]:
    Courbes(hebdo, variable, ncols=2, calendrier=calendrier, echelle=100, moyenne=True,
                     titre=titre + ' : au sein du Top 200', ylabel='Part (%)')
```

Les niveaux moyens donnent un repère (Top 5 des artistes : 37,7 % en Corée du Sud contre 16,2 % en Allemagne), et les courbes montrent les fluctuations autour de ces niveaux. Un pic ne suffit pas à identifier une sortie d'album : il faudrait examiner les titres et artistes de la semaine.

Trois choix de calcul sont testés : réintégrer les semaines exclues, attribuer toutes les écoutes d'une collaboration au premier artiste du crédit (au lieu d'un partage égal) et retenir les mêmes 134 semaines pour tous les pays. Si les résultats restent proches, les conclusions dépendent peu de ces choix.

```python
base['part'] = base['streams'] / base.groupby(cles)['streams'].transform('sum')
tous_top10 = base.sort_values(cles + ['rank']).groupby(cles).head(10).groupby(cles)['part'].sum().groupby('pays').mean()
verification = resume[['top10_titres', 'top5_artistes']].copy()
verification['top10_avec_incomplets'] = tous_top10

# Toutes les écoutes d'une collaboration attribuées au premier artiste du crédit (donnees['artiste'])
premiers = donnees.groupby(cles + ['artiste'])['streams'].sum().reset_index()
premiers['part'] = premiers['streams'] / premiers.groupby(cles)['streams'].transform('sum')
premiers = premiers.sort_values(cles + ['streams'], ascending=[True, True, False])
premiers['position'] = premiers.groupby(cles).cumcount() + 1
verification['top5_premier_credit'] = premiers[premiers['position'] <= 5].groupby(cles)['part'].sum().groupby('pays').mean()

# Mêmes semaines pour tous les pays
commun = hebdo[hebdo['semaine'].isin(semaines_communes)]
verification['top10_semaines_communes'] = commun.groupby('pays')['top10'].mean()
verification['top5_semaines_communes'] = commun.groupby('pays')['top5_artistes'].mean()
print((verification * 100).round(1).to_string())
```

Inclure les classements incomplets modifie la part moyenne du Top 10 d'au plus 0,61 point de pourcentage : son effet reste limité. En revanche, le traitement des collaborations peut changer sensiblement la concentration par artiste : en Grèce, la part des cinq premiers artistes passe de 28,2 % avec partage égal à 38,8 % avec attribution au premier crédit (+10,6 points). Les résultats par artiste doivent donc être présentés avec la convention utilisée.

**À retenir.** Les écoutes sont inégalement réparties sans être dominées par quelques chansons. La concentration apparaît surtout au niveau des artistes et dépend de la convention de partage des collaborations. Ces résultats décrivent le Top 200 et non l'ensemble des écoutes.

<h2 style="background:#eaf2fb;color:#12304d;padding:10px 18px;border-left:7px solid #eb6834;border-radius:0 8px 8px 0;font-size:21px;font-family:Georgia,serif;margin:22px 0 8px 0;"><a id="s4-2"></a>4.2. RESSEMBLANCE ENTRE LES CLASSEMENTS</h2>


La concentration décrit chaque classement pris seul : il reste à comparer les classements entre eux. Un même titre peut apparaître dans plusieurs pays : les dix pays comptent 23 000 titres distincts, mais le cumul des titres distincts de chaque pays atteint 28 885 (voir <a href="#s3-2">3.2</a>). Il existe donc des titres communs et des titres propres à chaque marché. Cette partie mesure d'abord la ressemblance entre les classements (<a href="#s4-2-1">4.2.1</a>), puis regroupe les pays selon cette ressemblance (<a href="#s4-2-2">4.2.2</a>).

<h3 style="color:#2a78d6;border-bottom:2px solid #2a78d6;padding-bottom:4px;font-size:17px;font-family:Georgia,serif;margin:18px 0 6px 0;"><a id="s4-2-1"></a>4.2.1. Similarités entre les classements</h3>


La concentration (<a href="#s4-1">4.1</a>) décrivait chaque classement pris seul. Ici, le point de vue est celui de deux classements entiers : à quel point se ressemblent-ils une même semaine ? Deux mesures sont calculées chaque semaine, puis moyennées :

- la **part de titres communs** (indice de Jaccard : nombre de titres présents dans les deux classements, divisé par le nombre de titres présents dans au moins un des deux) ;
- la **part des écoutes allant à ces titres communs**, moyenne des deux pays de la paire.

Deux hypothèses sont examinées. D'après <a href="#s3-2">3.2</a>, les pays aux classements les plus stables (Égypte, Japon, Thaïlande) devraient être ceux dont les classements ressemblent le moins aux autres. Par ailleurs, les pays qui partagent une langue devraient avoir des classements plus proches : seuls l'Égypte et le Maroc ont la même langue principale, l'arabe, avec des dialectes très différents.

```python
lignes = []
for semaine, tab in base.groupby("semaine"):
    titres_semaine = {p: set(g["uri"]) for p, g in tab.groupby("pays")}
    ecoutes = {p: g.groupby("uri")["streams"].sum() for p, g in tab.groupby("pays")}

    for a, b in combinations(pays_liste, 2):
        if a not in titres_semaine or b not in titres_semaine:
            continue
        communs = titres_semaine[a] & titres_semaine[b]
        union = titres_semaine[a] | titres_semaine[b]

        jaccard = len(communs) / len(union)
        part_a = ecoutes[a][list(communs)].sum() / ecoutes[a].sum()
        part_b = ecoutes[b][list(communs)].sum() / ecoutes[b].sum()

        lignes.append({"semaine": semaine, "pays_a": a, "pays_b": b,
                       "jaccard": jaccard, "part_ecoutes": (part_a + part_b) / 2})

sim = pd.DataFrame(lignes)
print(sim.shape)
print(sim[["jaccard", "part_ecoutes"]].describe().round(3))
```

```python
def matrice_similarite(indice):
    m = sim.groupby(["pays_a", "pays_b"])[indice].mean().unstack()
    m = m.reindex(index=pays_liste, columns=pays_liste)
    m = m.combine_first(m.T)                 # on recopie la moitié symétrique
    for p in pays_liste:
        m.loc[p, p] = np.nan                 # diagonale vide (un pays face à lui-même)
    return (m * 100).round(1)                # en pourcentage

legendes_indices = {"jaccard": "Titres communs (%), moyenne sur les semaines",
                    "part_ecoutes": "Écoutes allant à des titres communs (%), moyenne sur les semaines"}

for indice, legende in legendes_indices.items():
    carte_chaleur(matrice_similarite(indice), legende)
```

Les classements ressemblent peu : même la paire la plus proche, Allemagne et Portugal, ne partage que 15,7 % de ses titres, et la plupart des paires moins de 5 %. Les titres communs pèsent environ deux fois plus dans les écoutes que dans le nombre de titres (30,0 % des écoutes pour Allemagne et Portugal) : ce qui circule entre pays, ce sont surtout les grands succès internationaux.

Les paires les plus proches sont Allemagne et Portugal, Corée du Sud et Thaïlande, Corée du Sud et Japon, Allemagne et France, France et Portugal : la K-pop relie les pays asiatiques, et des succès européens relient les pays européens. Le Maroc est nettement plus proche de la France (16,6 % des écoutes) que de l'Égypte (5,0 %), alors qu'il partage avec elle l'arabe : ce qui rapproche les classements est la langue réellement partagée dans la musique, ici le français, et non la langue officielle. Enfin, l'Égypte et le Japon sont bien les plus éloignés des autres pays, ce qui confirme l'hypothèse issue de <a href="#s3-2">3.2</a>, à l'exception du lien du Japon avec la Corée du Sud.

<h3 style="color:#2a78d6;border-bottom:2px solid #2a78d6;padding-bottom:4px;font-size:17px;font-family:Georgia,serif;margin:18px 0 6px 0;"><a id="s4-2-2"></a>4.2.2. Groupes de pays</h3>


Les proximités de <a href="#s4-2-1">4.2.1</a> sont systématisées par une classification ascendante hiérarchique : les pays sont regroupés pas à pas, des plus proches aux plus éloignés. La distance entre deux pays est le complément de la part moyenne des écoutes allant à des titres communs (matrice de <a href="#s4-2-1">4.2.1</a>) : plus deux classements partagent d'écoutes, plus les pays sont proches.

```python
# 1. Matrice de distance : 1 - part moyenne des écoutes allant à des titres communs
sim_ecoutes = matrice_similarite("part_ecoutes") / 100
distance = 1 - sim_ecoutes.fillna(1)
dist = squareform(distance.to_numpy(), checks=False)
labels_pays = distance.index.tolist()

Z = linkage(dist, method="average")

# On suit les groupes formés à chaque étape pour afficher des noms de pays lisibles
composition = {i: [nom] for i, nom in enumerate(labels_pays)}
for etape, (g1, g2, hauteur, taille) in enumerate(Z, start=1):
    g1, g2 = int(g1), int(g2)
    composition[len(labels_pays) + etape - 1] = composition[g1] + composition[g2]
    print(f"Étape {etape} : {', '.join(composition[g1])}  +  {', '.join(composition[g2])}"
          f"  ->  distance {hauteur:.3f}")

# 2. Dendrogramme (les groupes sont colorés sous la hauteur de coupe)
fig, ax = plt.subplots(figsize=(9, 5))
dendrogram(Z, labels=labels_pays, color_threshold=0.85, leaf_rotation=45, ax=ax)
ax.set_ylabel("Distance (1 − part des écoutes communes)")
ax.set_title("Classification des pays selon la similarité de leurs classements")
ax.grid(False)
fig.tight_layout()
plt.show()
plt.close(fig)
```

Le dendrogramme (l'arbre des regroupements) se lit ainsi : la hauteur d'une fusion est la distance à laquelle deux groupes se rejoignent. Deux groupes se dégagent :

- **Groupe européen :** Allemagne et Portugal (0,700), puis la France (0,784) et le Maroc (0,876).
- **Groupe asiatique :** Corée du Sud et Thaïlande (0,746), puis le Japon (0,832).

La Grèce et l'Argentine rejoignent le groupe européen tardivement (0,914 et 0,943), et l'Égypte, qui rejoint les autres en dernier (0,967), est le pays le plus isolé. Le Maroc se rattache au groupe européen bien avant l'Égypte, ce qui confirme le constat de <a href="#s4-2-1">4.2.1</a>. Les distances restent toutefois élevées (au moins 0,70) : même au sein d'un groupe, les pays ne partagent qu'une part limitée de leurs écoutes.

**Conclusion de la partie <a href="#s4">4</a>.** Les classements des dix pays restent largement nationaux : les titres communs sont peu nombreux et se limitent pour l'essentiel aux grands succès internationaux. Leurs ressemblances suivent des logiques régionales, avec un bloc européen et un bloc asiatique, plutôt qu'une langue officielle commune. L'écoute sur Spotify ne conduit donc pas à une uniformisation des goûts, mais à la coexistence de répertoires nationaux et de quelques circulations régionales.

Les indicateurs `National`, `LangueLocale` et `Anglais` utilisés dans cette partie sont construits en <a href="#s2-4">2.4</a>. Le tableau `donnees` contient les classements complets retenus en <a href="#s4-1">4.1</a>.

<h3 style="color:#2a78d6;border-bottom:2px solid #2a78d6;padding-bottom:4px;font-size:17px;font-family:Georgia,serif;margin:18px 0 6px 0;"><a id="s5-1-1"></a>5.1.1. Part des répertoires nationaux dans les classements</h3>


Pour chaque pays et chaque semaine, la **part nationale** est le rapport entre les écoutes des titres nationaux et les écoutes totales du classement ; la même formule s'applique à la langue locale et à l'anglais. On fait ensuite la moyenne des semaines complètes, chacune pesant autant. Deux indicateurs complètent ces parts :

- **Couverture** : proportion des écoutes dont l'origine ou la langue est renseignée. Elle mesure la disponibilité de l'information, pas sa fiabilité.
- **Borne haute** : part identifiée à laquelle on ajoute les écoutes dont l'information manque, c'est-à-dire le cas où tous les titres non renseignés appartiendraient à la catégorie. Elle ne couvre pas les erreurs possibles dans les annotations connues.

Les calculs portent sur le Top 200 et sur le Top 50 ; le tableau affiché concerne le Top 200.

Les parts sont calculées semaine par semaine puis moyennées : chaque semaine pèse autant, quel que soit son volume d'écoutes. La part calculée sur l'ensemble de la période, qui pondère chaque semaine par son volume, s'en écarte d'au plus 1,6 point selon le pays (contrôle plus bas).

```python
# Calculer les parts pour chaque pays et semaine, puis pour le Top 50
indicateurs = ['National', 'LangueLocale', 'Anglais']
libelles = ['Origine nationale identifiée', 'Langue locale identifiée', 'Anglais identifié']
lignes = []
for top in [200, 50]:
    selection = donnees[donnees['rank'] <= top]
    for (pays, semaine), groupe in selection.groupby(['pays', 'semaine']):
        total = groupe['streams'].sum()
        ligne = {'pays': pays, 'semaine': semaine, 'top': top}
        for indicateur in indicateurs:
            ecoutes = groupe.loc[groupe[indicateur] == 1, 'streams'].sum()
            connu = 'langue_connue'
            if indicateur == 'National':
                connu = 'origine_connue'
            inconnues = groupe.loc[~groupe[connu], 'streams'].sum()
            ligne[indicateur] = ecoutes / total
            ligne[indicateur + '_borne_haute'] = (ecoutes + inconnues) / total
        ligne['couverture_origine'] = groupe.loc[groupe['origine_connue'], 'streams'].sum() / total
        ligne['couverture_langue'] = groupe.loc[groupe['langue_connue'], 'streams'].sum() / total
        lignes.append(ligne)

parts_hebdo = pd.DataFrame(lignes)
cols_parts = indicateurs + ['couverture_origine', 'couverture_langue']
moyennes = parts_hebdo.groupby(['top', 'pays'])[cols_parts].mean()
print((moyennes.loc[200] * 100).round(1).to_string())
```

En France, les titres d'origine nationale représentent en moyenne 69,5 % des écoutes du Top 200, contre 45,7 % en Allemagne (23,8 points d'écart). Au Portugal, la part nationale (24,8 %) est très inférieure à celle de la langue locale (51,2 %) : la langue ne se confond pas avec l'origine des artistes. En Corée du Sud, 76,9 % des écoutes relèvent d'artistes nationaux, 60,5 % de chansons en coréen et 48,5 % de chansons comportant de l'anglais, ces parts se recoupant puisqu'une chanson peut être bilingue. La couverture est de 100 % à l'arrondi dans chaque pays (seuls 2 langues et 6 nationalités ne sont pas renseignées, voir <a href="#s2-3">2.3</a>) : la borne haute est donc quasiment confondue avec la part identifiée.

Les moyennes masquent des semaines particulières : les courbes suivantes donnent l'évolution hebdomadaire de chaque part. La bande représente la marge liée aux champs inconnus, et non les erreurs possibles des annotations.

```python
# Évolution hebdomadaire de chaque part (Top 200), avec la borne haute en bande
courbes = parts_hebdo[parts_hebdo['top'] == 200]
for indicateur, titre in zip(indicateurs, libelles):
    Courbes(courbes, indicateur, ncols=2, calendrier=calendrier, echelle=100, ylim=(0, 100),
                     bande={indicateur: indicateur + '_borne_haute'}, ylabel='Part (%)',
                     titre=titre + ' : part des écoutes du Top 200',
                     note='Bande : marge liée aux champs inconnus, pas aux erreurs possibles des annotations.')
```

En Allemagne, la part nationale varie de 15,9 % (25 décembre 2025) à 70,3 % (31 août 2023), et la part d'anglais atteint 86,6 % la semaine du 25 décembre 2025 contre 55,6 % en moyenne : cette semaine est celle du pic de Noël identifié en <a href="#s3-1-3">3.1.3</a>, où les chansons de Noël (Wham!, Mariah Carey, Shakin' Stevens) dominent le Top 10 : cela est cohérent avec la chute de la part nationale et la hausse de l'anglais. Au Japon, la part nationale reste comprise entre 78,3 % et 95,7 %, ce qui traduit une présence nationale majoritaire sur toute la période. En France, elle varie de 51,8 % à 78,8 %.

**Contrôle sur le Top 50.** L'écart entre la part nationale du Top 50 et celle du Top 200, en points de pourcentage, indique si les origines nationales sont davantage représentées parmi les plus gros succès.

```python
verification = moyennes['National'].unstack('top') * 100
verification.columns = ['Top 50 (%)', 'Top 200 (%)']
verification['Écart Top 50 - Top 200 (points)'] = verification['Top 50 (%)'] - verification['Top 200 (%)']
print(verification.round(1).to_string())
```

En France, la part nationale passe de 69,5 % (Top 200) à 73,4 % (Top 50), soit +4,0 points, et de 76,9 % à 83,7 % en Corée du Sud (+6,8 points). L'écart est négatif en Argentine (−3,2 points) et en Thaïlande (−2,1 points). Le contraste France-Allemagne reste visible dans le Top 50 (73,4 % contre 48,6 %) : il ne dépend pas seulement des titres classés entre les rangs 51 et 200.

**Lien avec le renouvellement des classements (<a href="#s3-2">3.2</a>).** L'Égypte et le Japon, dont les classements sont les plus stables, sont aussi ceux où l'anglais pèse le moins (4,3 % et 9,9 % des écoutes) et où la langue locale domine (91,4 % et 86,8 %). La Thaïlande, également stable, s'en écarte (25,1 % d'anglais). À l'inverse, l'Allemagne, au renouvellement le plus rapide, a une part nationale faible (45,7 %) et 55,6 % d'anglais, mais la Grèce, elle aussi à renouvellement rapide, n'a que 8,6 % d'anglais. L'anglais n'explique donc pas à lui seul le rythme de renouvellement.

**Parts d'écoutes et parts de titres.** Une part d'écoutes mélange deux choses : le nombre de titres d'une catégorie et l'écoute moyenne de chacun. Pour séparer les deux, on compare, pour le Top 200, la **part des titres** (proportion des 200 places occupées par la catégorie, moyennée sur les semaines) à la part des écoutes. On calcule aussi le **rapport des médianes** : dans une semaine donnée, écoute médiane d'un titre de la catégorie divisée par celle d'un titre qui n'en est pas, puis moyenne sur les semaines (un rapport de 1,10 signifie qu'un titre de la catégorie est écouté 10 % de plus). La médiane est préférée à la moyenne, car quelques très gros succès tirent la moyenne vers le haut.

```python
# Part des titres, part des écoutes et rapport des médianes (Top 200, semaines complètes)
top200 = donnees[donnees['rank'] <= 200]
parts_top200 = parts_hebdo[parts_hebdo['top'] == 200]
blocs = []
for indicateur in ['National', 'LangueLocale']:
    # part des places du classement, puis moyenne des semaines
    part_titres = top200.groupby(['pays', 'semaine'])[indicateur].mean().groupby('pays').mean() * 100
    part_ecoutes = parts_top200.groupby('pays')[indicateur].mean() * 100
    # écoute médiane d'un titre de la catégorie (1) ou hors catégorie (0), chaque semaine
    medianes = top200.groupby(['pays', 'semaine', indicateur])['streams'].median().unstack()
    rapport = (medianes[1] / medianes[0]).groupby('pays').mean()
    # part des écoutes sur toute la période, pondérée par le volume de chaque semaine
    ecoutes_categorie = top200['streams'].where(top200[indicateur] == 1, 0)
    sur_periode = ecoutes_categorie.groupby(top200['pays']).sum() / top200.groupby('pays')['streams'].sum() * 100
    blocs.append(pd.DataFrame({'indicateur': indicateur, 'part_titres': part_titres,
                               'part_ecoutes': part_ecoutes, 'rapport_medianes': rapport,
                               'ecart_part_periode': sur_periode - part_ecoutes}))
comparaison_titres = pd.concat(blocs).reset_index().set_index(['indicateur', 'pays'])
print(comparaison_titres.round(2).to_string())
print("Écart absolu maximal entre la part sur la période et la moyenne des parts hebdomadaires :",
      round(comparaison_titres['ecart_part_periode'].abs().max(), 2), "point(s)")
```

La part des titres et la part des écoutes sont proches : dans la plupart des pays, l'écart est de 1 à 3 points. En Allemagne, 44,2 % des places du Top 200 sont occupées par des titres nationaux, qui reçoivent 45,7 % des écoutes ; en France 67,4 % et 69,5 % ; au Japon 86,2 % et 87,0 %. Le poids d'un répertoire vient donc surtout du **nombre** de titres qui le composent, et non d'une écoute plus forte de chacun d'eux. Le rapport des médianes le confirme : un titre d'origine nationale est écouté 4 % de plus qu'un titre étranger en France (rapport de 1,04) et 6 % de plus en Allemagne (1,06), et un titre en langue locale 3 % de plus dans ces deux pays (1,03).

Quelques pays s'écartent de ce constat. En Corée du Sud, les titres nationaux occupent 71,1 % des places mais reçoivent 76,9 % des écoutes (rapport de 1,21), et au Maroc 57,6 % des places pour 62,9 % des écoutes (1,17). En Grèce, les titres en grec occupent 87,4 % des places pour 91,3 % des écoutes (1,23). La Thaïlande est le seul pays où les titres en langue locale sont moins écoutés que les autres : 74,5 % des places, mais 68,5 % des écoutes (rapport de 0,94), les titres étrangers, notamment coréens, y étant très écoutés. Ces écarts sont descriptifs : on compare des titres déjà classés dans le Top 200, qui diffèrent aussi par leur popularité et leur promotion.

<h3 style="color:#2a78d6;border-bottom:2px solid #2a78d6;padding-bottom:4px;font-size:17px;font-family:Georgia,serif;margin:18px 0 6px 0;"><a id="s5-1-2"></a>5.1.2. Origine des écoutes dans chaque pays</h3>


La matrice donne, pour chaque pays d'écoute (colonne), la part moyenne des écoutes attribuée à chaque origine d'artiste (ligne). Pour éviter le double comptage, les écoutes d'un titre sont partagées également entre ses origines distinctes ; les origines inconnues forment une catégorie séparée et les origines minoritaires sont regroupées dans « autres ». Cette convention diffère de la part nationale précédente, où une collaboration compte intégralement dès qu'une origine est nationale : la diagonale de la matrice ne reproduit donc pas exactement cette part.

```python
# Qui écoute les musiques de quels pays ?
# Pour cette matrice seulement, on partage également les écoutes entre origines.
# Une collaboration FR/US reçoit moitié FR, moitié US : pas de double comptage.
repartition = donnees.copy()
origines_matrice = []
for origines in repartition['origines']:
    if len(origines) == 0:
        origines_matrice.append(['inconnue'])
    else:
        origines_matrice.append(origines)
repartition['origine'] = origines_matrice
repartition['ecoutes_partagees'] = repartition['streams'] / repartition['origine'].str.len()
repartition = repartition.explode('origine')
matrice_semaine = repartition.groupby(['pays', 'semaine', 'origine'])['ecoutes_partagees'].sum().unstack(fill_value=0)
matrice_semaine = matrice_semaine.div(matrice_semaine.sum(axis=1), axis=0)
matrice_origines = matrice_semaine.groupby('pays').mean().T
principales = list(matrice_origines.mean(axis=1).nlargest(15).index)
if 'inconnue' not in principales:
    principales.append('inconnue')
vue = matrice_origines.reindex(principales, fill_value=0)
vue.loc['autres'] = matrice_origines.drop(index=principales, errors='ignore').sum()
carte_chaleur(vue * 100, 'Origines des artistes et pays d’écoute (partage égal entre origines)',
               figsize=(12, 8), vmin=0, vmax=100, label='Part moyenne (%)',
               labels_y=vue.index.str.upper())
print((matrice_origines[["France", "Allemagne"]].loc[["fr", "de", "us", "gb"]] * 100).round(1).to_string())
```

L'origine américaine représente 31,4 % des écoutes en Allemagne contre 14,0 % en France, et l'origine britannique 10,0 % contre 4,8 %. Les origines locales dominent sur leur propre marché (origine française : 64,2 % en France contre 1,9 % en Allemagne ; origine allemande : 40,2 % en Allemagne contre 1,1 % en France) : la matrice fait apparaître à la fois un ancrage national et une forte présence de répertoires étrangers. Les 64,2 % sont inférieurs aux 69,5 % de part nationale parce que les collaborations internationales sont ici réparties entre leurs origines.

Le fichier des quotas classe les pays selon la présence indiquée d'un quota radio. Le scénario prudent oppose trois pays avec quota indiqué (Argentine, France, Portugal) à cinq pays avec absence indiquée (Allemagne, Égypte, Japon, Corée du Sud, Thaïlande).

Ce classement est appliqué à toute la période : il ne reconstitue pas les changements de lois, ni le niveau ou l'application des obligations, et les quotas concernent la radio alors que les résultats portent sur Spotify. Les absences indiquées ne sont pas toutes documentées, d'où l'expression « absence indiquée ».

La comparaison porte sur les semaines où les dix pays ont un classement complet. On calcule la moyenne de chaque pays, puis celle des pays de chaque groupe : chaque pays pèse autant, quel que soit son volume d'écoutes. Les colonnes `National`, `LangueLocale` et `Anglais` sont des proportions (0,489 correspond à 48,9 %), `scenario` distingue les deux classifications et `top` le périmètre (50 ou 200).

```python
# Rapprocher les quotas et les résultats
pays_par_code = {code: pays for pays, code in codes_pays.items()}
quotas['pays'] = quotas['country'].map(pays_par_code)
assert quotas['pays'].notna().all() and not quotas['pays'].duplicated().any()
assert set(quotas['pays']) == set(pays_liste)
assert quotas['quota'].isin([0, 1]).all()
quotas['groupe_fourni'] = quotas['quota'].map({1: 'Quota indiqué', 0: 'Absence indiquée'})
quotas['groupe_prudent'] = quotas['groupe_fourni']
# La source grecque ne démontre pas un quota radio obligatoire.
# Certains cahiers des charges marocains contiennent des obligations.
# Les autres absences sont déclarées dans le fichier, mais non vérifiées.
quotas.loc[quotas['country'].isin(['gr', 'ma']), 'groupe_prudent'] = 'Mis à part'
parts_hebdo = parts_hebdo.merge(quotas[['pays', 'groupe_fourni', 'groupe_prudent']],
                                on='pays', validate='many_to_one')

# Indicatrice de quota par pays, réutilisée en 5.2
quota_par_pays = quotas.set_index('pays')['quota']
print(quotas[["pays", "quota", "groupe_prudent"]].to_string(index=False))
```

```python
# Comparer sur les mêmes semaines (semaines_communes, voir 3.3), avec le même poids pour chaque pays
commun = parts_hebdo[parts_hebdo['semaine'].isin(semaines_communes)]
moyennes_pays = commun.groupby(['top', 'pays', 'groupe_fourni', 'groupe_prudent'])[indicateurs].mean().reset_index()
comparaisons = []
for scenario in ['groupe_fourni', 'groupe_prudent']:
    selection = moyennes_pays[moyennes_pays[scenario] != 'Mis à part']
    tableau = selection.groupby(['top', scenario])[indicateurs].mean()
    tableau['nombre_pays'] = selection.groupby(['top', scenario])['pays'].nunique()
    tableau = tableau.reset_index().rename(columns={scenario: 'groupe'})
    tableau['scenario'] = scenario
    comparaisons.append(tableau)
comparaison = pd.concat(comparaisons, ignore_index=True)
print("Semaines communes :", len(semaines_communes))
print(comparaison.round(3).to_string(index=False))
```

**Nature du quota : langue ou nationalité.** Tous les quotas ne portent pas sur la même chose. Selon les textes de loi, l'Argentine impose qu'au moins 30 % de la musique diffusée soit d'origine nationale, que ce soit par les auteurs ou par les interprètes (loi n° 26.522 de 2009) : c'est un quota de **nationalité**. La France impose 40 % de chansons d'expression française (loi du 1er février 1994, article 28 de la loi de 1986), et le Portugal 30 % de musique portugaise (loi n° 54/2010, modifiée par la loi n° 16/2024), définie par la langue portugaise ou le patrimoine culturel portugais, quelle que soit la nationalité des auteurs et des interprètes : ce sont des quotas de **langue**. La Grèce, avec la loi n° 5103/2024, impose 40 % de chansons en grec dans certains lieux publics, mais pas à la radio, et reste donc mise à part comme en <a href="#s5-1-3">5.1.3</a>. Cette distinction permet de comparer, pour chaque type de quota, la part que le quota vise : la part de langue locale pour les quotas de langue, la part nationale pour le quota de nationalité.

Le quota portugais est le seul à avoir varié pendant nos 210 semaines pour la radio. La loi sur la radio fixait une fourchette de 25 % à 40 % de musique portugaise ; un arrêté de 2021 l'avait portée à 30 %, mais il n'a pas été renouvelé début 2023. Si le quota agit sur ce que l'on écoute, la part de langue locale du Portugal pourrait donc monter après septembre 2023. La mesure grecque de 2024 concerne des lieux publics, pas la radio, et ne sert pas ici.

On compare la part des écoutes du Top 200 sur deux fenêtres de 13 semaines (environ trois mois) de part et d'autre du 1er septembre 2023, de juin à août 2023 puis de septembre à novembre 2023. Ces fenêtres évitent les semaines incomplètes du printemps 2023 (quelques semaines du Portugal n'ont que 198 ou 199 titres, et sont écartées comme dans le reste de l'étude) et la période de Noël, qui fait varier les parts. Le repère est la variation des cinq pays sans quota sur les mêmes fenêtres (chaque pays est comparé à lui-même, avec ses propres semaines complètes) : elle donne l'ordre de grandeur des mouvements d'une part d'un trimestre à l'autre en l'absence de quota.

```python
# Portugal avant et après le rétablissement du quota de 30 % (1er septembre 2023), Top 200
date_quota = pd.Timestamp('2023-09-01')
date_loi = pd.Timestamp('2024-02-05')
fenetre = pd.Timedelta(weeks=13)
temoins = ['Allemagne', 'Corée du Sud', 'Égypte', 'Japon', 'Thaïlande']   # pays sans quota
p200 = parts_hebdo[(parts_hebdo['top'] == 200) & parts_hebdo['pays'].isin(['Portugal'] + temoins)].copy()
p200['semaine'] = pd.to_datetime(p200['semaine'])
p200['periode'] = np.select([(p200['semaine'] >= date_quota - fenetre) & (p200['semaine'] < date_quota),
                             (p200['semaine'] >= date_quota) & (p200['semaine'] < date_quota + fenetre)],
                            ['Avant', 'Après'], default='Hors fenêtre')
fen = p200[p200['periode'] != 'Hors fenêtre']
print('Semaines complètes retenues par pays et par fenêtre :')
print(fen.groupby(['pays', 'periode'])['semaine'].nunique().unstack().to_string())

moy = fen.groupby(['pays', 'periode'])[['National', 'LangueLocale']].mean().mul(100).unstack('periode')
avant_apres = pd.DataFrame(index=['Portugal'] + temoins)
for indicateur in ['National', 'LangueLocale']:
    avant_apres[indicateur + ' avant'] = moy[indicateur]['Avant']
    avant_apres[indicateur + ' après'] = moy[indicateur]['Après']
    avant_apres[indicateur + ' variation'] = moy[indicateur]['Après'] - moy[indicateur]['Avant']
avant_apres.loc['Moyenne des 5 pays sans quota'] = avant_apres.loc[temoins].mean()
print(avant_apres.round(1).to_string())

for indicateur in ['LangueLocale', 'National']:
    variations = avant_apres.loc[temoins, indicateur + ' variation']
    print(indicateur, ": variation du Portugal", round(avant_apres.loc['Portugal', indicateur + ' variation'], 1),
          "point(s) ; pays sans quota de", round(variations.min(), 1), "à", round(variations.max(), 1), "point(s)")
```

<h2 style="background:#eaf2fb;color:#12304d;padding:10px 18px;border-left:7px solid #eb6834;border-radius:0 8px 8px 0;font-size:21px;font-family:Georgia,serif;margin:22px 0 8px 0;"><a id="s5-2"></a>5.2. INDICE DE BIAIS DOMESTIQUE</h2>


Pour chaque pays, on compare deux parts de musique nationale : sa part dans le Top 200 de son propre pays, et sa part dans les Top 200 des neuf autres pays.

indice = part de musique nationale dans son pays ÷ part de cette musique dans les 9 autres pays

Si l'indice est plus grand que 1, les gens écoutent plus leur musique nationale que les autres pays ne le font. Les parts sont calculées avec les écoutes (streams).

Une chanson est « nationale » si au moins un de ses artistes est du pays.

```python
lignes_indice = []

for nom, code in codes_pays.items():
    chez_lui = base[base["pays"] == nom]
    autres = base[base["pays"] != nom]
    nat_chez_lui = chez_lui["origines"].apply(lambda o: code in o)
    nat_ailleurs = autres["origines"].apply(lambda o: code in o)
    part_chez_lui = chez_lui.loc[nat_chez_lui, "streams"].sum() / chez_lui["streams"].sum()
    part_ailleurs = autres.loc[nat_ailleurs, "streams"].sum() / autres["streams"].sum()
    lignes_indice.append([nom, part_chez_lui, part_ailleurs, part_chez_lui / part_ailleurs])

indices = pd.DataFrame(lignes_indice, columns=["pays", "part_chez_lui", "part_ailleurs", "indice"])
print(indices.round(6))
```

Les indices vont de 21 à plus de 45 000. On utilise une échelle logarithmique pour tout voir sur le même graphique.

```python
couleurs = {"quota": "tab:blue", "pas de quota": "lightgray"}
indices["groupe"] = indices["pays"].map(quota_par_pays).map({1: "quota", 0: "pas de quota"})

plt.figure(figsize=(8, 5))
plt.barh(indices["pays"], indices["indice"], color=indices["groupe"].map(couleurs))
plt.xscale("log")
valeurs_dans_barres(plt.gca(), fmt="{:.0f}")
plt.xlabel("Indice de biais domestique (échelle logarithmique)\nbleu : pays avec quota, gris : pas de quota")
plt.title("Indice de biais domestique par pays")
plt.tight_layout()
plt.show()
plt.close()
```

**Ce que montre l'indice**

- Dans les 10 pays, la musique nationale pèse beaucoup dans le Top 200 du pays : de 25 % (Portugal) à 87 % (Japon) des écoutes. Les indices sont tous bien plus grands que 1.
- Les indices sont très différents : de 21 (Corée du Sud) à plus de 45 000 (Portugal). Du plus haut au plus bas : Portugal, Égypte, Thaïlande, Grèce, Argentine, Maroc, Japon, Allemagne, France, Corée du Sud.
- On ne voit pas de séparation claire entre pays avec quota et sans quota : le Portugal (quota) est premier, la Grèce (quota) quatrième, l'Argentine (quota) cinquième, la France (quota) neuvième, et l'Égypte et la Thaïlande (sans quota) ont des indices très élevés.

Il n'y a pas non plus de séparation nette entre les deux groupes : parmi les pays avec quota, la part nationale va de 24,8 % (Portugal) à 69,5 % (France), et parmi les pays sans quota de 45,7 % (Allemagne) à 87,0 % (Japon). L'indice de biais domestique ne sépare pas non plus les groupes (<a href="#s5-2">5.2</a>) : le Portugal, la Grèce et l'Argentine ont des indices élevés, mais l'Égypte et la Thaïlande, sans quota, aussi, et la France, avec quota, est parmi les plus bas.

**La musique nationale et la langue locale pèsent beaucoup, avec ou sans quota.** Au Japon, 87,0 % des écoutes du Top 200 vont à des titres d'au moins un artiste japonais, contre 24,8 % au Portugal (<a href="#s5-1-1">5.1.1</a>). Les pays sans quota ne sont pas les moins tournés vers leur musique : le Japon fait 87,0 %, l'Égypte 81,3 % et la Corée du Sud 76,9 %. La langue locale pèse encore plus lourd dans plusieurs pays (Argentine 93,6 %, Égypte 91,4 %, Grèce 91,3 %). Au Portugal, la part de langue locale (51,2 %) est le double de la part nationale (24,8 %) : la langue d'une chanson et la nationalité de l'artiste ne se confondent pas. L'indice de biais domestique (<a href="#s5-2">5.2</a>) est supérieur à 1 dans les dix pays, de 21 en Corée du Sud à plus de 45 000 au Portugal. Un indice très élevé signale surtout que cette musique est presque absente des classements des autres pays.

**Ce poids vient du nombre de titres, pas d'une écoute individuelle beaucoup plus forte.** Un titre d'origine nationale est écouté en médiane 4 % de plus en France et 6 % de plus en Allemagne que les autres titres du même classement la même semaine (<a href="#s5-1-1">5.1.1</a>). Les écarts plus nets sont des exceptions : Corée du Sud (+21 %), Maroc (+17 %), et Grèce pour la langue locale (+23 %).

**Part nationale et diversité sont deux choses distinctes.** Au Japon, la part nationale est de 87,0 % et les cinq premiers artistes réunissent 36,3 % des écoutes (<a href="#s4-1-2">4.1.2</a>) ; en Allemagne, la part nationale est de 45,7 % et les cinq premiers artistes ne réunissent que 16,2 %. Dans chaque pays, les dix premières chansons représentent de 14,7 % à 21,2 % des écoutes, mais la concentration par artiste varie beaucoup : une forte présence de musique nationale ne dit donc rien de la diversité des artistes écoutés.

**Les classements se ressemblent peu.** En moyenne, deux pays n'ont que 3,5 % de titres en commun (<a href="#s4-2-1">4.2.1</a>). Les paires les plus proches sont Allemagne et Portugal (15,7 %), puis des paires régionales : Corée du Sud et Thaïlande, Corée du Sud et Japon (<a href="#s4-2-2">4.2.2</a>). Les titres communs sont surtout de grands succès internationaux.

**Les trajectoires dans le temps ne s'alignent pas sur les quotas.** Le total du Top 200 monte puis baisse à partir de 2024 ou 2025 en France, au Portugal, en Grèce et en Argentine (<a href="#s3-1">3.1</a>), mais l'Allemagne, sans quota, baisse aussi (−24,2 %), alors que l'Égypte (+512 %), la Corée du Sud (+272 %), le Maroc (+128 %) et la Thaïlande (+100 %) montent fortement.

Les pays les plus tournés vers leur musique sont le Japon, l'Égypte et la Corée du Sud. Les plus ouverts sont le Portugal (24,8 %) et l'Allemagne (45,7 %), qui écoutent aussi beaucoup d'anglais (38,2 % et 55,6 % des écoutes, <a href="#s5-1-1">5.1.1</a>).

<h2 style="background:#eaf2fb;color:#12304d;padding:10px 18px;border-left:7px solid #eb6834;border-radius:0 8px 8px 0;font-size:21px;font-family:Georgia,serif;margin:22px 0 8px 0;"><a id="s6-2"></a>6.2 D'autres explications possibles</h2>


- **La langue.** Le Japon et la Thaïlande ont une langue que personne d'autre ne parle dans notre échantillon, et l'Égypte ne la partage qu'avec le Maroc. Leurs classements changent peu (<a href="#s3-2">3.2</a>), et le Japon et l'Égypte écoutent très peu d'anglais (9,9 % et 4,3 %).

- **L'anglais.** L'Allemagne (55,6 %) et le Portugal (38,2 %) écoutent beaucoup d'anglais, et ce sont les pays avec le moins de musique nationale. Ce n'est pas vrai partout : la Corée du Sud a 48,5 % d'anglais (<a href="#s5-1-1">5.1.1</a>) et 76,9 % de musique nationale.

- **La taille du marché.** Le Top 200 allemand fait environ 128 à 169 millions d'écoutes par semaine (fin et début de période). Celui de l'Égypte en fait 7 à 43 millions et celui du Maroc 6 à 14 millions. Dans un petit marché, un seul artiste peut tout changer : au Maroc, une semaine, ElGrandeToto fait 53 % du Top 200.

- **La proximité entre nationalité et langue.** Dans plusieurs pays, la part nationale et la part de langue locale sont presque égales (Japon 87,0 % et 86,8 %, Allemagne 45,7 % et 41,2 %) : on ne peut pas y distinguer ce qui tient à l'artiste et ce qui tient à la langue. Elles divergent au Portugal (24,8 % et 51,2 %) et en Argentine (57,9 % et 93,6 %).

- **Spotify.** Notre base ne contient que le Top 200 de Spotify, pas toute l'écoute d'un pays. Les playlists et les recommandations de la plateforme influencent aussi les titres visibles. Nous n'avons pas pu tester cette piste.

<h2 style="background:#eaf2fb;color:#12304d;padding:10px 18px;border-left:7px solid #eb6834;border-radius:0 8px 8px 0;font-size:21px;font-family:Georgia,serif;margin:22px 0 8px 0;"><a id="s6-3"></a>6.3 Pourquoi on ne conclut pas sur l'effet des quotas</h2>


1. Les quotas concernent la radio, et nous n'avons que des écoutes sur Spotify. Nous ne voyons pas la radio.
2. Les quotas existaient déjà avant le début de nos données (France 1986, Portugal 2001, Argentine 2009), et la comparaison « avant » et « après » n'est donc possible que pour peu de mouvements. Deux seulement ont lieu pendant notre période : le rétablissement du quota portugais à 30 % en septembre 2023, étudié en <a href="#s5-1-5">5.1.5</a>, et la loi grecque de 2024, qui concerne des lieux publics et non la radio.
3. Les pays avec et sans quota sont très différents : la langue, la taille du marché. Nous n'avons que dix pays.
4. Le résultat change selon la façon de classer la Grèce et le Maroc (54,3 % contre 69,0 % d'un côté, 50,0 % contre 70,2 % de l'autre).
5. Un pays met peut-être un quota parce que sa musique nationale est faible, comme le Portugal. Le quota serait alors la conséquence, pas la cause.
6. Les quotas ne sont pas tous les mêmes : 30 % en Argentine, 40 % en France et en Grèce, 30 % aujourd'hui au Portugal. Leur nature diffère aussi (langue ou nationalité), mais chaque type ne concerne que peu de pays : deux pays pour la langue, un seul pour la nationalité.
7. Nous comparons des parts moyennes de pays, pas l'évolution d'un même pays avec et sans quota : les résultats décrivent des écarts, pas des effets.

<h2 style="background:#eaf2fb;color:#12304d;padding:10px 18px;border-left:7px solid #eb6834;border-radius:0 8px 8px 0;font-size:21px;font-family:Georgia,serif;margin:22px 0 8px 0;"><a id="s6-4"></a>6.4 Ce que l'étude apporte malgré tout</h2>


- Nous avons comparé dix pays sur la même période : 210 semaines et 419 891 lignes, en distinguant la nationalité des artistes et la langue des chansons.
- Au Portugal, le rétablissement du quota à 30 % en septembre 2023 ne s'accompagne d'aucun changement de la part de langue locale (−1,1 point, dans la fourchette des pays sans quota).
- Les pays avec quota n'ont pas une part plus élevée de la catégorie que leur quota vise : la langue locale pour la France et le Portugal (61,1 % contre 70,0 % sans quota), la musique nationale pour l'Argentine (55,6 % contre 70,2 %).
- Les quotas ne sont pas nécessaires pour avoir une forte musique nationale : le Japon, l'Égypte et la Corée du Sud n'ont pas de quota dans notre fichier et sont les plus tournés vers leur musique.
- Le poids de la musique nationale vient surtout du nombre de titres : dans six pays sur dix, un titre national est écouté en médiane 4 à 7 % de plus que les autres (<a href="#s5-1-1">5.1.1</a>).
- Les écoutes se concentrent surtout sur quelques artistes, bien plus que sur quelques chansons (<a href="#s4-1">4.1</a>), et les classements restent différents d'un pays à l'autre : deux pays n'ont en moyenne que 3,5 % de titres en commun (<a href="#s4-2">4.2</a>).
- Pour aller plus loin : avoir des données de radio, et regarder avant et après la mesure de la Grèce (2024).

<h1 style="background:linear-gradient(90deg,#1f4e79,#2a78d6);color:#ffffff;padding:14px 22px;border-radius:10px;font-size:26px;font-family:Georgia,serif;margin:28px 0 10px 0;border:none;"><a id="conclusion"></a>CONCLUSION <a href="#plan" style="float:right;font-size:12px;color:#dbe8f7;text-decoration:none;font-family:sans-serif;">↑ retour au plan</a></h1>


Ce travail cherchait à savoir si les quotas de diffusion influencent la répartition des écoutes entre artistes nationaux et non nationaux, et entre chansons en langue locale et dans une autre langue. **Dans nos données, on ne voit pas cet effet.** Les pays avec quota n'ont pas une part plus élevée de la catégorie que leur quota vise. Pour les quotas de langue (France, Portugal), la part de langue locale est de 61,1 % en moyenne, contre 70,0 % dans les cinq pays sans quota. Pour le quota de nationalité (Argentine), la part nationale est de 55,6 %, contre 70,2 %. Toutes catégories confondues, les pays avec quota ont même moins d'écoutes nationales que les autres (54,3 % contre 69,0 %), et l'indice de biais domestique ne sépare pas les deux groupes. Le seul « avant et après » possible dans nos données, le rétablissement du quota portugais à 30 % en septembre 2023, ne montre aucune rupture de la part de langue locale (62,4 % avant, 61,3 % après). Les pays les plus tournés vers leur musique, comme le Japon (87,0 % d'écoutes nationales) ou l'Égypte (81,3 %), n'ont pas de quota dans notre fichier.

Il ne faut pas en conclure que les quotas sont inefficaces. Ils concernent la radio, alors que nous observons Spotify. La plupart existaient avant le début de nos données, ce qui limite les comparaisons « avant » et « après », et les pays diffèrent par leur langue, la taille de leur marché et leur histoire musicale. Un pays peut aussi avoir adopté un quota parce que sa musique nationale était faible. Ce que les données permettent de dire est plus limité : **rien dans la répartition des écoutes Spotify ne justifie d'attribuer aux quotas la place de la musique nationale ou de la langue locale.** Cette place est forte partout, avec ou sans quota, et vient surtout du nombre de titres nationaux présents dans les classements, plus que d'une écoute individuelle nettement supérieure.

Les analyses préalables situent ce résultat. Les classements ont des dynamiques et des rythmes de renouvellement très différents d'un pays à l'autre. Les écoutes se concentrent sur quelques artistes plus que sur quelques chansons. Les classements de deux pays ont peu de titres en commun, et ceux-ci sont surtout de grands succès internationaux, avec des ressemblances régionales (un bloc européen, un bloc asiatique). Une étude causale demanderait des données de radio et un historique réglementaire vérifié.
